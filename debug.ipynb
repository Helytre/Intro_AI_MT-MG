```python
import pandas
from Dependency.File_converter import selecting

from Dependency.one_hot_encoding import one_hot_encoding

# 1. récupérer le split habituel
data_training, data_testing = selecting()

# 2. appliquer le one-hot
result = one_hot_encoding(data_training, data_testing)
oh_training = result["training"]
oh_testing = result["testing"]

# 3. les vérifications
print("Taille avant :", data_training.shape, data_testing.shape)
print("Taille après :", oh_training.shape, oh_testing.shape)
print("Mêmes colonnes train/test :", list(oh_training.columns) == list(oh_testing.columns))
print("Mêmes lignes qu'avant :", oh_training.index.equals(data_training.index), oh_testing.index.equals(data_testing.index))
print("Valeurs vides :", oh_training.isna().sum().sum(), oh_testing.isna().sum().sum())
print("Colonne Unnamed présente :", [c for c in oh_training.columns if c.startswith("Unnamed")])

# 4. regarder à quoi ça ressemble
# oh_training.filter(like="Weather_Conditions").head()

pandas.set_option("display.max_columns", None)   # aucune limite sur le nombre de colonnes
pandas.set_option("display.width", None)          # ne pas couper selon la largeur
oh_training
```
