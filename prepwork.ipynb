```python
import pandas

#reading uncleaned dataset

dataset = pandas.read_csv("Dataset/UK_Accident.csv")

#understanding dataset

pandas.set_option('display.max_columns', 35)
dataset.head()
```

```python
#counting all class / value in most colunms

test = dataset.columns.to_list()
delete = ["Unnamed: 0", "Unnamed: 0.1", "Accident_Index","Location_Easting_OSGR","Location_Northing_OSGR","Longitude","Latitude","LSOA_of_Accident_Location"]
for i in delete:
    if i in test:
        test.remove(i)
for val in test:
    print(dataset[val].value_counts())
    print("\n")
```

```python
#seeing variables relationships

for val in test:
    print(dataset.groupby(val)["Accident_Severity"].mean())
    print("\n")
```

```python
#same as before but with balanced accident_severity class

dataset = dataset.groupby('Accident_Severity').sample(n=10000, random_state=1)

for val in test:
    print(dataset.groupby(val)["Accident_Severity"].mean())
    print("\n")
```
